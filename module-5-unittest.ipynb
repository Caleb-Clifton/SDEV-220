## Testing Your Code

Unit testing is used to check whether a small piece of code works the way it is expected to work. Instead of manually checking the result every time, a test can compare the actual result from a function to the expected result. If the actual result matches the expected result, the test passes. If not, the test fails and shows that something needs to be fixed.

## Writing Your First Test

The tutorial first created a package folder that contained an `__init__.py` file. Inside this file, a custom `sum()` function was written to add together the values from a list.

````python
# __init__.py

def sum(arg):
    total = 0
    for val in arg:
        total += val
    return total
````

The test file was named `test.py`. It created a test class that inherited from `unittest.TestCase`. Inside that class, the method `test_list_int()` tested whether the sum of `[1, 2, 3]` returned `6`.

## Test Results Explanation

The test result means that the function worked as expected for the input that was tested. The single dot means one test was run successfully. The line `Ran 1 test` confirms that Python’s unittest module found and executed one test method. The `OK` message means the expected result and actual result matched.
