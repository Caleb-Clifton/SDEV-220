## 13.1

Write the current date as a string to the text file today.txt.

```python
from datetime import date

today = date.today().isoformat()

with open("today.txt", "w") as file:
    file.write(today)
```

## 13.2

Read the text file today.txt into the string today_string.

```python
with open("today.txt", "r") as file:
    today_string = file.read()

today_string
```

## 13.3

Parse the date from today_string.

```python
from datetime import datetime

parsed_date = datetime.strptime(today_string, "%Y-%m-%d").date()

parsed_date
```

## 15.1

Use multiprocessing to create three separate processes. Make each one wait a random number of seconds between zero and one, print the current time, and then exit.

I'll be making a script to save as a .py file so the notebook can call it below.

```python
%%writefile multiprocessing_time.py

import multiprocessing
import random
import time
from datetime import datetime


def print_time(start_time):
    time.sleep(random.random())

    elapsed_ms = int((time.perf_counter() - start_time) * 1000)

    print(f"{datetime.now().strftime("%H:%M:%S.%f")[:-3]} | +{elapsed_ms} ms")


if __name__ == "__main__":
    start_time = time.perf_counter()

    print("Start time:", datetime.now().strftime("%H:%M:%S.%f")[:-3], flush=True)

    processes = []

    for _ in range(3):
        p = multiprocessing.Process(target=print_time, args=(start_time,))
        processes.append(p)
        p.start()

    for p in processes:
        p.join()
```
